# Plotting Correlation Heatmap (Polars)

```python
"""
Code from 1a_read_arff_polars.ipynb has been saved to .py file named
get_dataset_polars.py.

The get_dataset_polars.get_polars_df() function simply returns a Polars
dataframe.
"""
from get_dataset_polars import get_polars_df
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Gets Polars dataframe
raw = get_polars_df()
```

```python
# Correlation heatmap
# Jupyter
(raw
 .to_pandas(use_pyarrow_extension_array=True)
 .corr()
 .style.background_gradient(cmap='RdBu', vmin=-1, vmax=1)
)
```

```python
# Correlation heatmap
# Python IDLE equivalent
# Converts Polars datafram to Pandas for plotting.
df = raw.to_pandas()
corr = df.corr(numeric_only=True)
plt.figure(figsize=(20, 10))
sns.heatmap(corr, cmap="RdBu", vmin=-1, vmax=1, annot=True) #-1 will be red, 1 will be blue
plt.title("Correlation Heatmap")
plt.show()
```
